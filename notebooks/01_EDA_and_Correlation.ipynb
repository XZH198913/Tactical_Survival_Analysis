# 探索性数据分析与基准相关性（系统领土的制图学）

## 1. 理论引言：资本系统的基础制图
在探寻特洛伊木马的逃逸线之前，我们必须首先理解蓝筹画廊系统是如何运作并维持其稳态的。本节通过探索性数据分析（EDA）对数据全貌进行制图。我们将计算各核心变量（伪装质量、不可消化性、核心稳定性）的基础分布，并构建相关性热力图。
这些线性相关性构成了系统的“常态”——它揭示了常规策略下的运作逻辑。只有看清了这种强烈的同化力量，我们才能在后续章节中理解“非线性（倒 U 型）”的病理学裂隙。

## 2. 数据加载与系统地貌总览
载入清洗后的宽表数据，查看缺失值、数据类型以及核心指标的描述性统计特征。

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 绘图审美配置
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid", palette="muted")

# 载入结构化数据表
df = pd.read_csv("micropolitical_diagnosis_table.csv")

# 提取核心定量指标
core_metrics = ['Camouflage_Quality', 'Indigestibility_Score', 'Core_Stability']
df_core = df[core_metrics].dropna()

print(f"✅ 数据地貌勘测完成。有效样本量 N = {len(df_core)}")
display(df_core.describe().round(3))
```

**地貌诠释**
描述性统计为我们提供了系统的边界参数。观察 Camouflage_Quality 的中位数与极值，可以初步判断大部分艺术作品在面对画廊资本时，选择了何种程度的伪装妥协。这构成了系统最基础的“平均审美”。

## 3. 多维变量的线性纠缠：相关性热力图
变量之间并非孤立存在。我们计算皮尔逊（Pearson）相关系数矩阵，以揭示资本病理学中的基础纠缠状态。

```python
# 计算皮尔逊相关系数矩阵
corr_matrix = df_core.corr()

# 绘制下三角热力图（去除重复冗余信息）
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt=".3f", cmap="coolwarm", 
            vmax=1, vmin=-1, center=0, square=True, linewidths=.5, cbar_kws={"shrink": .75})

plt.title("基准变量的皮尔逊相关性热力图", fontsize=14, fontweight='bold', pad=20)
plt.show()
```

**哲学合成：线性的幻觉**
如果在热力图中，我们观察到“伪装质量”与“不可消化性”之间存在微弱的线性正相关或零相关，这其实是一种统计学上的幻象。这种全局的线性计算掩盖了局部区间的激烈对抗。这也是为什么我们不能止步于此，必须在下一章（Notebook 1）中引入二次回归，去寻找被线性关系抹平的“倒 U 型”真相。

## 4. 初探突围：全局线性散点分布
为了直观展示上述线性统计的局限性，我们绘制伪装质量与不可消化性的二维散点图，并拟合一条基础的线性回归线。

```python
plt.figure(figsize=(10, 6))
sns.regplot(
    data=df_core, 
    x='Camouflage_Quality', 
    y='Indigestibility_Score',
    scatter_kws={'alpha':0.5, 's':50, 'edgecolor':'w'},
    line_kws={'color':'red', 'linestyle':'--', 'linewidth':2},
    label='全局线性拟合 (Global Linear Fit)'
)

plt.title("全局线性分布：伪装与不可消化性的表面关系", fontsize=14, fontweight='bold')
plt.xlabel("伪装质量 (Camouflage Quality)", fontsize=12)
plt.ylabel("不可消化性 (Indigestibility Score)", fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()

# 提取并打印线性相关的 P 值
from scipy import stats
r, p = stats.pearsonr(df_core['Camouflage_Quality'], df_core['Indigestibility_Score'])
print(f"全局线性皮尔逊相关系数 r = {r:.3f}, p值 = {p:.4f}")
```

**结论导向下一章**
当图表呈现在眼前时，我们会发现红色的线性拟合线（虚线）显得极其生硬，大量高分样本游离于这条线段之外。系统深层的张力拒绝被简单的直线捕获。
